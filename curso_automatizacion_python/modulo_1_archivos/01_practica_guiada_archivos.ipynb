```python
from pathlib import Path
from tempfile import TemporaryDirectory

import shutil
```

## 1. Crear un entorno seguro

`TemporaryDirectory` crea una carpeta que se elimina al terminar el bloque.

```python
with TemporaryDirectory() as carpeta_temporal:
    carpeta = Path(carpeta_temporal)
    (carpeta / 'reporte.pdf').write_text('contenido', encoding='utf-8')
    (carpeta / 'datos.csv').write_text('nombre,valor\nA,10', encoding='utf-8')
    (carpeta / 'notas.txt').write_text('recordatorio', encoding='utf-8')
    print(sorted(archivo.name for archivo in carpeta.iterdir()))
```

## 2. Inspeccionar rutas

Observa que `Path` permite preguntar si una ruta existe, si es archivo y cuál es su extensión.

```python
with TemporaryDirectory() as carpeta_temporal:
    carpeta = Path(carpeta_temporal)
    archivo = carpeta / 'reporte.final.pdf'
    archivo.write_text('contenido', encoding='utf-8')
    assert archivo.exists()
    assert archivo.is_file()
    assert archivo.suffix == '.pdf'
    print(archivo.name, archivo.stem, archivo.suffix)
```
